```c++
# Define the three numbers
num1 = 15
num2 = 34
num3 = 21

# Method 1: Using the built-in max() function
largest = max(num1, num2, num3)
print(f"The largest number is: {largest}")


# Method 2: Using conditional (if-elif-else) statements
if num1 >= num2 and num1 >= num3:
    largest_conditional = num1
elif num2 >= num1 and num2 >= num3:
    largest_conditional = num2
else:
    largest_conditional = num3

print(f"The largest number (using conditions) is: {largest_conditional}")
```
