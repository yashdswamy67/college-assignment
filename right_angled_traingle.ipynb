```c++
import math

def is_right_triangle(a, b, c):
    """
    Checks if three sides can form a right-angled triangle.
    Sorts sides so the largest side is automatically the hypotenuse.
    """
    sides = sorted([a, b, c])
    
    # Check triangle inequality theorem (sides must form a valid triangle)
    if sides[0] + sides[1] <= sides[2] or sides[0] <= 0:
        return False
    
    # Safely handle floating-point precision issues
    return math.isclose(sides[0]**2 + sides[1]**2, sides[2]**2)

def main():
    # Define numerical values directly for a, b, and c
    a = 3
    b = 4
    c = 5

    print("--- Right-Angled Triangle Checker ---")
    print(f"Side a = {a}")
    print(f"Side b = {b}")
    print(f"Side c = {c}")

    # Pass the variables to the function
    if is_right_triangle(a, b, c):
        print(f"\nResult: Yes, ({a}, {b}, {c}) forms a right-angled triangle!")
    else:
        print(f"\nResult: No, ({a}, {b}, {c}) does NOT form a right-angled triangle.")

if __name__ == "__main__":
    main()
```
