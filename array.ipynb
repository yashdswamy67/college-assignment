```c++
import re

def is_alphanumeric_regex(s: str) -> bool:
    # ^ starting anchor
    # [a-zA-Z0-9]+ matches one or more allowed characters
    # $ ending anchor
    pattern = r'^[a-zA-Z0-9]+$'
    return bool(re.match(pattern, s))

# Examples
test_strings = ["Hello123", "Python_3", "User@2026", "SimpleText", ""]

for text in test_strings:
    print(f"'{text}': {is_alphanumeric_regex(text)}")
```
